```python
for _ in range(3):
    img_dilated = morphology.binary_dilation(img_dilated)
img_closed = morphology.remove_small_holes(img_dilated, area_threshold=100)

skeleton = morphology.skeletonize(img_closed)
pruned_skeleton, _, _ = pcv.morphology.prune(skeleton.astype('uint8'), 0)
```
